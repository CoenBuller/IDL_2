```python
import matplotlib.pyplot as plt

# Carica dataset
X = np.load("fractal_images.npy")
y = np.load("fractal_labels.npy")

fig, axes = plt.subplots(4, 3, figsize=(9, 12))
fig.suptitle("Esempi per ciascun tipo di frattale (3 per classe)", fontsize=14, y=0.92)

# Per ogni tipo (0=Dream, 1=Hopalong, 2=De Jong, 3=Clifford)
for label in range(4):
    # Trova i primi 3 indici corrispondenti a questo tipo
    indices = np.where(y == label)[0][:3]
    
    for col, idx in enumerate(indices):
        axes[label, col].imshow(X[idx])
        axes[label, col].set_title(f"Label: {label}", fontsize=10)
        axes[label, col].axis('off')

plt.tight_layout()
plt.show()
```
